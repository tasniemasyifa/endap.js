```
import random
import numpy as np
import matplotlib.pyplot as plt

# Parameter simulasi
N = 100000

# Batas area pembatas (Bounding Box) untuk Gambar Kiri
x_min, x_max = -1, 1
y_min, y_max = -1, 2

# List untuk menyimpan koordinat titik
x_irisan, y_irisan = [], []
x_luar, y_luar = [], []

# Proses Simulasi Monte Carlo
for _ in range(N):
    x_acak = random.uniform(x_min, x_max)
    y_acak = random.uniform(y_min, y_max)

    # Syarat berada di dalam KEDUA lingkaran:
    # Lingkaran 1 (Pusat 0,0): x^2 + y^2 <= 1
    # Lingkaran 2 (Pusat 0,1): x^2 + (y-1)^2 <= 1
    l1 = x_acak**2 + y_acak**2 <= 1
    l2 = x_acak**2 + (y_acak - 1)**2 <= 1

    if l1 and l2:
        x_irisan.append(x_acak)
        y_irisan.append(y_acak)
    else:
        x_luar.append(x_acak)
        y_luar.append(y_acak)

# Perhitungan Luas Kotak Pembatas (2 x 3 = 6)
luas_kotak = (x_max - x_min) * (y_max - y_min)

# Estimasi Luas Irisan Monte Carlo
estimasi_luas = (len(x_irisan) / N) * luas_kotak

# Nilai Eksak Matematika untuk Irisan 2 Lingkaran r=1, d=1
# Luas = 2*(pi/3) - sqrt(3)/2
luas_eksak = (2 * np.pi / 3) - (np.sqrt(3) / 2)

# Cetak Hasil Perhitungan
print("=== ESTIMASI MONTE CARLO (GAMBAR KIRI) ===")
print(f"Estimasi Luas Irisan (Hijau) : {estimasi_luas:.4f}")
print(f"Nilai Eksak Matematika       : {luas_eksak:.4f}")

# Visualisasi Plot Tunggal
plt.figure(figsize=(6, 7))

# Plot Titik-Titik
plt.scatter(x_luar, y_luar, color='lightgray', s=1, label='Di Luar Irisan')
plt.scatter(x_irisan, y_irisan, color='seagreen', s=1, label='Daerah Irisan (Hijau)')

# Garis Kurva Lingkaran
theta = np.linspace(0, 2 * np.pi, 200)
plt.plot(np.cos(theta), np.sin(theta), color='red', linestyle='--', label='x² + y² = 1')
plt.plot(np.cos(theta), 1 + np.sin(theta), color='blue', linestyle='--', label='x² + (y-1)² = 1')

# Pengaturan Grafik
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.gca().set_aspect('equal', adjustable='box')

plt.title(f"Monte Carlo Simulation - Irisan 2 Lingkaran\nEstimasi Luas = {estimasi_luas:.4f} (Eksak ≈ {luas_eksak:.4f})")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

```
import random
import numpy as np
import matplotlib.pyplot as plt

# Parameter simulasi
N = 100000

# Area Pembatas (Bounding Box)
x_min, x_max = -1.0, 1.0
y_min, y_max = -1.5, 1.5

# List titik koordinat
x_hijau, y_hijau = [], []
x_biru, y_biru = [], []
x_oranye, y_oranye = [], []
x_luar, y_luar = [], []

for _ in range(N):
    x_acak = random.uniform(x_min, x_max)
    y_acak = random.uniform(y_min, y_max)

    # 1. Lingkaran Utama (Wajah Panda): Pusat (0,0), r = 1
    lingkaran_utama = (x_acak**2 + y_acak**2) <= 1

    # 2. Lingkaran Mata Biru (Depan/Utuh): Pusat (-0.5, 0), r = 0.5
    mata_biru = ((x_acak + 0.5)**2 + y_acak**2) <= 0.25

    # 3. Lingkaran Pembentuk Hijau: Pusat (1,1) dan (1,-1), r = 1
    lingkaran_hijau_atas  = ((x_acak - 1)**2 + (y_acak - 1)**2) <= 1
    lingkaran_hijau_bawah = ((x_acak - 1)**2 + (y_acak + 1)**2) <= 1

    # Hierarki Penentuan Warna:
    if lingkaran_utama:
        if mata_biru:
            # Mata biru berada di paling depan (utuh)
            x_biru.append(x_acak)
            y_biru.append(y_acak)
        elif lingkaran_hijau_atas or lingkaran_hijau_bawah:
            # Daerah hijau di sebelah kanan
            x_hijau.append(x_acak)
            y_hijau.append(y_acak)
        else:
            # Sisa area lingkaran utama menjadi oranye/cokelat
            x_oranye.append(x_acak)
            y_oranye.append(y_acak)
    else:
        x_luar.append(x_acak)
        y_luar.append(y_acak)

# Luas Kotak Pembatas (2 x 3 = 6)
luas_kotak = (x_max - x_min) * (y_max - y_min)

# Perhitungan Estimasi Luas
luas_biru = (len(x_biru) / N) * luas_kotak
luas_hijau = (len(x_hijau) / N) * luas_kotak
luas_oranye = (len(x_oranye) / N) * luas_kotak
total_luas = luas_biru + luas_hijau + luas_oranye

# Tampilkan Hasil Perhitungan
print("=== HASIL ESTIMASI MONTE CARLO ===")
print(f"Luas Mata Biru (Utuh)  : {luas_biru:.4f} (Eksak pi/4 ≈ {np.pi/4:.4f})")
print(f"Luas Daerah Hijau      : {luas_hijau:.4f}")
print(f"Luas Daerah Oranye     : {luas_oranye:.4f}")
print(f"Total Luas Lingkaran   : {total_luas:.4f} (Eksak pi ≈ {np.pi:.4f})")

# Visualisasi Plot
plt.figure(figsize=(7, 7))

plt.scatter(x_luar, y_luar, color='lightgray', s=1, label='Luar Lingkaran')
plt.scatter(x_oranye, y_oranye, color='chocolate', s=1, label=f'Oranye/Cokelat (L ≈ {luas_oranye:.3f})')
plt.scatter(x_hijau, y_hijau, color='seagreen', s=1, label=f'Hijau (L ≈ {luas_hijau:.3f})')
plt.scatter(x_biru, y_biru, color='steelblue', s=1, label=f'Biru Mata (L ≈ {luas_biru:.3f})')

# Garis Kurva Pembentuk untuk Memperjelas
theta = np.linspace(0, 2*np.pi, 200)
plt.plot(np.cos(theta), np.sin(theta), color='black', linewidth=1.2, label='Lingkaran Utama')
plt.plot(-0.5 + 0.5*np.cos(theta), 0.5*np.sin(theta), color='blue', linewidth=1)
plt.plot(1 + np.cos(theta), 1 + np.sin(theta), color='green', linestyle='--', linewidth=0.8)
plt.plot(1 + np.cos(theta), -1 + np.sin(theta), color='green', linestyle='--', linewidth=0.8)

plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.gca().set_aspect('equal', adjustable='box')

plt.title("Simulasi Monte Carlo - Wajah Panda (Revisi)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

```
import random
import numpy as np
import matplotlib.pyplot as plt

# Parameter Simulasi
N = 100000

# ============================================================
# 1. ANALISIS GAMBAR KIRI (Irisan 2 Lingkaran Miring)
# ============================================================
x_min1, x_max1 = -1.0, 2.0
y_min1, y_max1 = -1.0, 2.0
luas_kotak1 = (x_max1 - x_min1) * (y_max1 - y_min1)  # 3x3 = 9

x_irisan1, y_irisan1 = [], []

for _ in range(N):
    x_acak = random.uniform(x_min1, x_max1)
    y_acak = random.uniform(y_min1, y_max1)

    l1 = (x_acak**2 + y_acak**2) <= 1
    l2 = ((x_acak - 1)**2 + (y_acak - 1)**2) <= 1

    if l1 and l2:
        x_irisan1.append(x_acak)
        y_irisan1.append(y_acak)

estimasi_kiri = (len(x_irisan1) / N) * luas_kotak1
eksak_kiri = (np.pi / 2) - 1

error_mutlak_kiri = abs(estimasi_kiri - eksak_kiri)
persen_error_kiri = (error_mutlak_kiri / eksak_kiri) * 100


# ============================================================
# 2. ANALISIS GAMBAR KANAN (Wajah Panda & Komponen-Komponennya)
# ============================================================
x_min2, x_max2 = -1.0, 1.0
y_min2, y_max2 = -1.5, 1.5
luas_kotak2 = (x_max2 - x_min2) * (y_max2 - y_min2)  # 2x3 = 6

x_biru2, y_biru2 = [], []
x_hijau2, y_hijau2 = [], []
x_oranye2, y_oranye2 = [], []

for _ in range(N):
    x_acak = random.uniform(x_min2, x_max2)
    y_acak = random.uniform(y_min2, y_max2)

    lingkaran_utama = (x_acak**2 + y_acak**2) <= 1
    mata_biru       = ((x_acak + 0.5)**2 + y_acak**2) <= 0.25
    hijau_atas      = ((x_acak - 1)**2 + (y_acak - 1)**2) <= 1
    hijau_bawah     = ((x_acak - 1)**2 + (y_acak + 1)**2) <= 1

    if lingkaran_utama:
        if mata_biru:
            x_biru2.append(x_acak)
            y_biru2.append(y_acak)
        elif hijau_atas or hijau_bawah:
            x_hijau2.append(x_acak)
            y_hijau2.append(y_acak)
        else:
            x_oranye2.append(x_acak)
            y_oranye2.append(y_acak)

# Estimasi Luas Gambar Kanan
est_biru = (len(x_biru2) / N) * luas_kotak2
est_hijau = (len(x_hijau2) / N) * luas_kotak2
est_oranye = (len(x_oranye2) / N) * luas_kotak2
est_total_kanan = est_biru + est_hijau + est_oranye

# Nilai Eksak Total Lingkaran Utama (pi * r^2 = pi * 1^2 = pi)
eksak_kanan_total = np.pi
eksak_biru = np.pi * (0.5**2)  # pi/4 ≈ 0.7854

error_mutlak_kanan = abs(est_total_kanan - eksak_kanan_total)
persen_error_kanan = (error_mutlak_kanan / eksak_kanan_total) * 100


# ============================================================
# TAMPILKAN HASIL CETAK / RUNNING
# ============================================================
print("==================================================")
print("           ANALISIS ERROR GAMBAR KIRI             ")
print("==================================================")
print(f"Estimasi Luas Monte Carlo : {estimasi_kiri:.6f}")
print(f"Nilai Eksak (pi/2 - 1)     : {eksak_kiri:.6f}")
print(f"Error Mutlak              : {error_mutlak_kiri:.6f}")
print(f"Persentase Error          : {persen_error_kiri:.4f}%\n")

print("==================================================")
print("           ANALISIS ERROR GAMBAR KANAN            ")
print("==================================================")
print(f"Luas Mata Biru (Estimasi) : {est_biru:.4f}  (Eksak ≈ {eksak_biru:.4f})")
print(f"Luas Hijau (Estimasi)     : {est_hijau:.4f}")
print(f"Luas Oranye (Estimasi)    : {est_oranye:.4f}")
print(f"Total Luas Lingkaran Utama : {est_total_kanan:.6f}")
print(f"Nilai Eksak Total (pi)    : {eksak_kanan_total:.6f}")
print(f"Error Mutlak Total        : {error_mutlak_kanan:.6f}")
print(f"Persentase Error Total    : {persen_error_kanan:.4f}%")
print("==================================================")
```
